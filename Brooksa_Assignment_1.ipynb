# Assignment 1 — Madhava-Leibniz π Calculation

## Problem
I used the Madhava-Leibniz π example from class and changed it to test different numbers of iterations.

I defined **reasonable accuracy** as an absolute error less than **0.001**.


```python
import math
import matplotlib.pyplot as plt
```

## Part 1 — Test Different Numbers of Iterations


```python
N_values = [10, 100, 1000, 10000]
pi_values = []
errors = []

for N in N_values:
    pi = 0

    for n in range(1, N+1):
        if n == 1:
            term = 1
        else:
            term = -((-1)**n) / (2*n - 1)

        pi = pi + term

    pi = 4 * pi
    error = abs(pi - math.pi)

    pi_values.append(pi)
    errors.append(error)

    print("N =", N, "| pi =", pi, "| error =", error)
```

```python
plt.plot(N_values, errors, marker='o')
plt.xlabel("Number of iterations")
plt.ylabel("Absolute error")
plt.title("Error in the Madhava-Leibniz Approximation")
plt.show()
```

```python
N = 2
error = 1

while error >= 0.001:
    pi = 0

    for n in range(1, N+1):
        if n == 1:
            term = 1
        else:
            term = -((-1)**n) / (2*n - 1)

        pi = pi + term

    pi = 4 * pi
    error = abs(pi - math.pi)
    N = N + 1

print("Reasonable accuracy is reached at N =", N - 1)
print("Error =", error)
```

The approximation reaches my definition of reasonable accuracy at about **N = 1000**.


## Part 2 — Function


```python
def calculate_pi(N):
#Approximate pi using the Madhava-Leibniz series.

    pi = 0

    for n in range(1, N+1):
        if n == 1:
            term = 1
        else:
            term = -((-1)**n) / (2*n - 1)

        pi = pi + term

    return 4 * pi
```

```python
print(calculate_pi(10))
print(calculate_pi(100))
print(calculate_pi(1000))
```

Functions are useful because they let us reuse the same calculation with different inputs without rewriting the code.
